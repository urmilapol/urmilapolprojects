```python
# 1. Install prerequisites & Devanagari fonts
!apt-get install -y fonts-deva fonts-deva-extra libraqm-dev -qq
!pip install Pillow -q

from PIL import Image, ImageDraw, ImageFont
import os
from google.colab import files

# 2. Upload your horizontal Vishnu template image
print("Upload your horizontal template image:")
uploaded = files.upload()
img_name = list(uploaded.keys())[0]

# 3. Setup Devanagari Font
font_path = "/usr/share/fonts/truetype/fonts-deva-extra/gargi.ttf"
if not os.path.exists(font_path):
    font_path = "/usr/share/fonts/truetype/lohit-devanagari/Lohit-Devanagari.ttf"

# 4. Text Content for Shloka 3
shloka_title = "॥ श्लोक ५ चा भावार्थ ॥"
sanskrit_shlok =" स्वयम्भूः शम्भुरादित्यः पुष्कराक्षो महास्वनः । अनादिनिधनो धाता विधाता धातुरुत्तमः ॥"
marathi_meaning = "जे स्वतःहून प्रकट झालेले आहेत, भक्तांचे परम कल्याण करणारे आहेत, सूर्याचे मूळ तेज आहेत, कमळासारखे सुंदर नयन असलेले, वेदवाणीचे धनी, जन्म व मृत्यू नसलेले (अमर), विश्वाचा आधार, कर्मांनुसार फळ देणारे आणि सर्वांत श्रेष्ठ पोषणकर्ता आहेत."

# 5. Open Base Image & Calculate Dimensions
base_img = Image.open(img_name).convert("RGBA")
W, H = base_img.size

# Responsive Font Sizes
font_title = ImageFont.truetype(font_path, int(H * 0.065))
font_shlok = ImageFont.truetype(font_path, int(H * 0.052))
font_meaning = ImageFont.truetype(font_path, int(H * 0.046))

# Colors
COLOR_TITLE = (130, 25, 10)       # Deep Vedic Maroon
COLOR_SHLOK = (35, 20, 15)        # Dark Antique Brown
COLOR_MEANING = (20, 25, 30)      # Crisp Charcoal Ink
COLOR_ACCENT = (195, 145, 45)     # Warm Gold Line

# Text Area Placement (Right Half of the image)
right_start_x = int(W * 0.54)
margin_right = int(W * 0.95)
content_width = margin_right - right_start_x

# Text Wrapping Function
def wrap_text_pixels(text, font, max_width, draw):
    lines = []
    paragraphs = text.split('\n')
    for p in paragraphs:
        words = p.strip().split()
        if not words:
            continue
        cur_line = []
        for word in words:
            test_line = " ".join(cur_line + [word])
            bbox = draw.textbbox((0, 0), test_line, font=font, language='hi')
            if (bbox[2] - bbox[0]) <= max_width:
                cur_line.append(word)
            else:
                if cur_line:
                    lines.append(" ".join(cur_line))
                cur_line = [word]
        if cur_line:
            lines.append(" ".join(cur_line))
    return lines

# Bold Drawing Function
def draw_bold(draw, pos, text, font, fill, align_center=False, max_w=None):
    x, y = pos
    if align_center and max_w:
        bbox = draw.textbbox((0, 0), text, font=font, language='hi')
        text_w = bbox[2] - bbox[0]
        x = x + (max_w - text_w) // 2
    draw.text((x, y), text, font=font, fill=fill, language='hi')
    draw.text((x + 1, y), text, font=font, fill=fill, language='hi')
    draw.text((x, y + 1), text, font=font, fill=fill, language='hi')

# 6. Render Card
card = base_img.copy()
draw = ImageDraw.Draw(card)

# Vertical Cursor Start
y = int(H * 0.14)

# A. Title
draw_bold(draw, (right_start_x, y), shloka_title, font_title, COLOR_TITLE, align_center=True, max_w=content_width)
y += int(H * 0.08)

# Decorative Divider Line
line_w = int(content_width * 0.6)
line_x1 = right_start_x + (content_width - line_w) // 2
draw.line([(line_x1, y), (line_x1 + line_w, y)], fill=COLOR_ACCENT, width=2)
y += int(H * 0.06)

# B. Sanskrit Shloka (Centered)
shlok_lines = wrap_text_pixels(sanskrit_shlok, font_shlok, content_width, draw)
for line in shlok_lines:
    draw_bold(draw, (right_start_x, y), line, font_shlok, COLOR_SHLOK, align_center=True, max_w=content_w if 'content_w' in locals() else content_width)
    y += int(H * 0.07)

y += int(H * 0.04)  # Section Spacer

# C. Marathi Meaning (Left-aligned & neatly wrapped)
meaning_lines = wrap_text_pixels(marathi_meaning, font_meaning, content_width, draw)
for line in meaning_lines:
    draw_bold(draw, (right_start_x, y), line, font_meaning, COLOR_MEANING)
    y += int(H * 0.065)

# 7. Save and Download
output_file = "vishnu_shloka_5_landscape.png"
card.convert("RGB").save(output_file, "PNG", quality=95)
print(f"\nGenerated: {output_file}")
files.download(output_file)
```
