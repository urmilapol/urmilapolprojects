```python
# 1. Install required library
!pip install openpyxl -q

import pandas as pd
from google.colab import files
from openpyxl.styles import Alignment, Font, PatternFill
from openpyxl.utils import get_column_letter

# 2. Upload your CSV file to Google Colab
print("Please upload your CSV file:")
uploaded = files.upload()

# Get the uploaded filename automatically
csv_filename = list(uploaded.keys())[0]
print(f"\nUploaded: {csv_filename}")

# 3. Read the CSV using pandas with UTF-8 encoding
df = pd.read_csv(csv_filename, encoding='utf-8')

# Display first few rows to verify Devanagari text
display(df.head())

# 4. Write to Excel with formatting and column auto-sizing
output_excel = "Vishnu_Sahasranamam_Formatted.xlsx"

with pd.ExcelWriter(output_excel, engine='openpyxl') as writer:
    df.to_excel(writer, index=False, sheet_name='Vishnu Sahasranamam')

    workbook = writer.book
    worksheet = writer.sheets['Vishnu Sahasranamam']

    # Define styles
    header_fill = PatternFill(start_color="1F4E79", end_color="1F4E79", fill_type="solid")
    header_font = Font(name="Calibri", size=11, bold=True, color="FFFFFF")
    data_font = Font(name="Calibri", size=10)
    wrap_alignment = Alignment(wrap_text=True, vertical="top")
    center_alignment = Alignment(horizontal="center", vertical="top")

    # Format headers
    for col_num, col_name in enumerate(df.columns, 1):
        cell = worksheet.cell(row=1, column=col_num)
        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(horizontal="center", vertical="center")

    # Format data cells (Text wrapping & proper vertical alignment)
    for row in worksheet.iter_rows(min_row=2, max_row=len(df) + 1, min_col=1, max_col=len(df.columns)):
        for idx, cell in enumerate(row):
            cell.font = data_font
            if idx == 0:  # Shloka Number column
                cell.alignment = center_alignment
            else:
                cell.alignment = wrap_alignment

    # Set custom readable column widths
    worksheet.column_dimensions['A'].width = 18
    worksheet.column_dimensions['B'].width = 50
    worksheet.column_dimensions['C'].width = 80

print(f"\nExcel file successfully generated: {output_excel}")

# 5. Trigger download in your browser
files.download(output_excel)
```
