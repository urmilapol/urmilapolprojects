```python
# 1. Install prerequisites & Devanagari fonts
!apt-get install -y fonts-deva fonts-deva-extra libraqm-dev -qq
!pip install Pillow pandas -q

import os
from PIL import Image, ImageDraw, ImageFont
from google.colab import files

# 2. Upload your Vishnu color template image (vishnu1.jpg)
print("Upload the Lord Vishnu color image:")
uploaded = files.upload()
img_name = list(uploaded.keys())[0]

# 3. Setup Devanagari Font
font_path = "/usr/share/fonts/truetype/fonts-deva-extra/gargi.ttf"
if not os.path.exists(font_path):
    font_path = "/usr/share/fonts/truetype/lohit-devanagari/Lohit-Devanagari.ttf"

# 4. Text from Image
shloka_num = "१"
sanskrit_shloka = "विश्वं विष्णुर्वषट्कारो भूतभव्यभवत्प्रभुः ।\nभूतकृद्भूतभृद्भावो भूतात्मा भूतभावनः ॥"
marathi_meaning = "जगाचे रूप, सर्वव्यापक, यज्ञाचे प्रतीक, भूत-भविष्य-वर्तमान यांचे स्वामी, सर्व भूतांची निर्मिती आणि पोषण करणारे भगवान विष्णू आहेत."

# 5. Open Base Image & Calculate Scaling
base_img = Image.open(img_name).convert("RGBA")
W, H = base_img.size

# Scaled typography sizes based on resolution (adjusted for user request)
font_title = ImageFont.truetype(font_path, int(W * 0.06))
font_shlok = ImageFont.truetype(font_path, int(W * 0.05))
font_meaning = ImageFont.truetype(font_path, int(W * 0.04))

# Define Colors (adjusted for user request)
COLOR_TITLE = (150, 75, 0)        # Lighter Brown
COLOR_SHLOK = (200, 50, 50)      # More vibrant Red
COLOR_TEXT = (50, 0, 100)        # Dark Purple

# Text Wrapping Function
def wrap_text_pixels(text, font, max_width, draw):
    lines = []
    paragraphs = text.split('\n')
    for p in paragraphs:
        words = p.strip().split()
        if not words:
            continue
        cur_line = []
        for word in words:
            test_line = " ".join(cur_line + [word])
            bbox = draw.textbbox((0, 0), test_line, font=font, language='hi')
            if (bbox[2] - bbox[0]) <= max_width:
                cur_line.append(word)
            else:
                if cur_line:
                    lines.append(" ".join(cur_line))
                cur_line = [word]
        if cur_line:
            lines.append(" ".join(cur_line))
    return lines

# Bold Drawing Function
def draw_bold(draw, pos, text, font, fill, align_center=False, max_w=None):
    x, y = pos
    if align_center and max_w:
        bbox = draw.textbbox((0, 0), text, font=font, language='hi')
        text_w = bbox[2] - bbox[0]
        x = x + (max_w - text_w) // 2
    draw.text((x, y), text, font=font, fill=fill, language='hi')
    draw.text((x + 1, y), text, font=font, fill=fill, language=language_val)
    draw.text((x, y + 1), text, font=font, fill=fill, language=language_val)

language_val = 'hi'

# 6. Render Card
card = base_img.copy()
draw = ImageDraw.Draw(card)

# Positioning on lower parchment (adjusted for user request - move text further down)
margin_x = int(W * 0.11)
content_w = W - (2 * margin_x)
y = int(H * 0.75) # Start text rendering closer to the bottom

# 1. Title Heading
title_text = f"श्लोक {shloka_num} चा भावार्थ:"
draw_bold(draw, (margin_x, y), title_text, font=font_title, fill=COLOR_TITLE)
y += int(W * 0.06 * 1.5) # Adjusted line height for new font size

# 2. Sanskrit Shloka (Centered with deep maroon ink)
shlok_lines = sanskrit_shloka.split('\n')
for line in shlok_lines:
    draw_bold(draw, (margin_x, y), line.strip(), font=font_shlok, fill=COLOR_SHLOK, align_center=True, max_w=content_w)
    y += int(W * 0.05 * 1.4) # Adjusted line height for new font size

y += int(W * 0.02) # Spacer

# 3. Marathi Meaning
meaning_lines = wrap_text_pixels(marathi_meaning, font_meaning, content_w, draw)
for line in meaning_lines:
    draw_bold(draw, (margin_x, y), line, font=font_meaning, fill=COLOR_TEXT)
    y += int(W * 0.04 * 1.38) # Adjusted line height for new font size

# Save & Download Output
output_path = "vishnu_shlok_1_card.png"
card.convert("RGB").save(output_path, "PNG", quality=95)
print(f"\nSuccessfully generated {output_path}!")

files.download(output_path)
```
