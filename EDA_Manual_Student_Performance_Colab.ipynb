# Mini Project 1: Exploratory Data Analysis (EDA)

**Topic:** Student Performance Analysis  
**Platform:** Google Colab  

This notebook demonstrates manual EDA using Pandas, NumPy, Matplotlib and Seaborn.

## Learning Objectives
- Load and inspect a dataset
- Detect missing values and duplicates
- Perform descriptive statistics
- Create visualizations
- Study correlations
- Perform binning and grouping
- Interpret EDA findings

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
print('Libraries imported successfully.')
```

## 1. Create the Dataset

For this classroom mini project, we create a small student-performance dataset.

```python
data = {
    'Student': ['Asha','Rahul','Sneha','Amit','Priya','Vijay','Neha','Rohan','Pooja','Kiran','Anita','Suresh'],
    'Gender': ['Female','Male','Female','Male','Female','Male','Female','Male','Female','Male','Female','Male'],
    'Study_Hours': [5,3,6,2,7,4,8,3,6,5,7,2],
    'Attendance': [90,75,95,60,92,80,98,70,88,85,94,65],
    'Marks': [88,65,92,55,95,72,98,60,85,78,91,58]
}

df = pd.DataFrame(data)
df
```

## 2. Basic Data Inspection

```python
print('First 5 rows:')
display(df.head())

print('Shape:', df.shape)
print('Columns:', df.columns.tolist())

print('\nData types and non-null values:')
df.info()
```

## 3. Check Missing Values and Duplicates

```python
print('Missing values:')
print(df.isnull().sum())

print('\nNumber of duplicate rows:', df.duplicated().sum())
```

## 4. Descriptive Statistics

```python
display(df.describe())

print('Average marks:', df['Marks'].mean())
print('Highest marks:', df['Marks'].max())
print('Lowest marks:', df['Marks'].min())
```

## 5. Gender Distribution

```python
print(df['Gender'].value_counts())

sns.countplot(data=df, x='Gender')
plt.title('Number of Students by Gender')
plt.show()
```

## 6. Distribution of Marks

```python
sns.histplot(data=df, x='Marks', kde=True)
plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.show()
```

## 7. Study Hours vs Marks

```python
sns.scatterplot(data=df, x='Study_Hours', y='Marks', hue='Gender', s=100)
plt.title('Study Hours vs Marks')
plt.show()
```

## 8. Attendance vs Marks

```python
sns.scatterplot(data=df, x='Attendance', y='Marks', hue='Gender', s=100)
plt.title('Attendance vs Marks')
plt.show()
```

## 9. Correlation Matrix

```python
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()
display(corr)

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 10. Pair Plot

```python
sns.pairplot(df, vars=['Study_Hours', 'Attendance', 'Marks'], hue='Gender')
plt.show()
```

## 11. Binning Marks into Performance Categories

```python
bins = [0, 40, 60, 75, 100]
labels = ['Fail', 'Pass', 'First Class', 'Distinction']

df['Performance'] = pd.cut(df['Marks'], bins=bins, labels=labels, include_lowest=True)
display(df[['Student', 'Marks', 'Performance']])

sns.countplot(data=df, x='Performance')
plt.title('Performance Categories')
plt.xticks(rotation=20)
plt.show()
```

## 12. Grouping and Aggregation

```python
gender_summary = df.groupby('Gender', observed=True)[['Study_Hours', 'Attendance', 'Marks']].mean()
display(gender_summary)

performance_summary = df.groupby('Performance', observed=True)['Marks'].agg(['count', 'mean'])
display(performance_summary)
```
